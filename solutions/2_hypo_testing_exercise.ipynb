# Hypothesis Testing Exercise: Solution

A worked reference for the two case studies, following the 7-step framework:

1. **Understand the data type**: continuous or discrete.
2. **Identify the analytical question**: determine which distribution fits your question.
3. **Calculate descriptive statistics**: compute and visualise summary statistics.
4. **Interpret the graph and descriptive statistics**: analyse the data's distribution and patterns.
5. **Check for normality**: assess whether the data fits a normal distribution.
6. **Apply transformations or non-parametric tests**: adjust or select methods based on normality.
7. **Formulate and test hypotheses**: conduct the hypothesis test that answers the research question.

## Case Study 1: Evaluating Employee Training Effectiveness

**Business context.** A company implemented a new training programme to enhance employee skills. After the training, they tested 20 randomly selected employees.

The company wants to evaluate whether the new training programme led to a significant increase in the average test score compared to the historical average score of 70.

> [!TIP]
> "A significant increase" is a directional claim. Think about what that means for your alternative hypothesis before you reach for a test.

```python
import pandas as pd

data = {
    "Test_Score": [
        75,
        80,
        78,
        72,
        85,
        77,
        70,
        88,
        90,
        73,
        82,
        74,
        76,
        79,
        83,
        81,
        69,
        71,
        87,
        84,
    ]
}
df_e_t = pd.DataFrame(data)
```

### Step 1: Understand the Data Type

`Test_Score` is a numeric measurement on a continuous scale (scores can take many values across a wide range), so we treat it as **continuous** data. That points us towards the normal-distribution family of tests rather than counts or proportions.

### Step 2: Identify the Analytical Question

Did the training programme **increase** the average test score above the historical average of 70?

The claim is directional ("a significant increase"), so the alternative hypothesis is one-sided:

- H0: mu = 70 (the average score is unchanged)
- H1: mu > 70 (the average score has increased)

### Step 3: Calculate Descriptive Statistics

```python
# Summary statistics for the training scores
df_e_t["Test_Score"].describe()
```

The sample mean is about 78.7 with a standard deviation of about 6.2, and every value lies between 69 and 90. The mean sits well above the historical 70, so there is something worth testing.

### Step 4: Interpret the Graph and Descriptive Statistics

```python
import matplotlib.pyplot as plt

plt.hist(df_e_t["Test_Score"], bins=6, alpha=0.7, color="blue")
plt.title("Test Scores After Training")
plt.xlabel("Test Score")
plt.ylabel("Frequency")
plt.show()
```

The histogram is roughly symmetric around the high 70s with no obvious outliers. With only 20 observations a histogram is an unreliable normality check on its own, so we verify properly in the next step.

### Step 5: Check for Normality

```python
import scipy.stats as stats

# Q-Q plot: points close to the line indicate approximate normality
stats.probplot(df_e_t["Test_Score"], dist="norm", plot=plt)
plt.title("Q-Q Plot of Test Scores")
plt.show()

# Shapiro-Wilk: H0 is that the data is normal, so a large p-value is good news
_, p_shapiro = stats.shapiro(df_e_t["Test_Score"])
print(f"Shapiro-Wilk p-value: {p_shapiro:.3f}")
```

### Step 6: Transformation or Non-Parametric Test?

The Q-Q plot points lie close to the diagonal and the Shapiro-Wilk p-value is about 0.75, far above 0.05, so there is no evidence against normality. No transformation or non-parametric test is needed: a **one-sample t-test** is appropriate (t rather than z because the population standard deviation is unknown).

### Step 7: Formulate and Test Hypotheses

```python
# One-sided one-sample t-test: does the mean exceed the historical 70?
t_stat, p_value = stats.ttest_1samp(df_e_t["Test_Score"], 70, alternative="greater")
print(f"T-statistic: {t_stat:.3f}, p-value: {p_value:.2e}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: the training increased average scores.")
else:
    print("Fail to reject the null hypothesis: no evidence of an increase.")
```

**Explanation**

The t-statistic is about 6.23 with a one-sided p-value of about 2.7e-06, far below 0.05, so we reject H0: the training programme raised the average test score significantly above the historical 70. The test is one-sided because the business claim is directional ("led to a significant increase"); `alternative='greater'` puts all of alpha in the upper tail. A two-sided test would simply double the p-value (about 5.5e-06) and would reach the same conclusion here, but matching the test to the claim is the correct practice.

## Case Study 2: Analysing Average Weekly Hours Worked

**Business context.** A company wants to assess whether employees are working more or fewer hours than the standard 40 hours per week. The HR department has collected the number of hours worked by a random sample of 25 employees during a typical week.

The company wants to determine whether the average number of hours worked deviates significantly from the standard.

> [!TIP]
> "More or fewer" is not the same claim as "a significant increase" in Case Study 1. Let that difference show up in your hypotheses.

```python
import pandas as pd

# Data
data = {
    "Hours_Worked": [
        42,
        38,
        40,
        41,
        39,
        45,
        43,
        37,
        40,
        44,
        39,
        40,
        42,
        41,
        38,
        40,
        39,
        43,
        38,
        40,
        42,
        37,
        39,
        40,
        41,
    ]
}
df = pd.DataFrame(data)
```

### Step 1: Understand the Data Type

`Hours_Worked` is a numeric measurement of time, so we treat it as **continuous** data.

### Step 2: Identify the Analytical Question

Does the average number of hours worked **deviate** from the 40-hour standard, in either direction?

"More or fewer" is a non-directional claim, so unlike Case Study 1 the alternative is two-sided:

- H0: mu = 40 (average hours match the standard)
- H1: mu != 40 (average hours differ from the standard)

### Step 3: Calculate Descriptive Statistics

```python
# Summary statistics for weekly hours
df["Hours_Worked"].describe()
```

The sample mean is about 40.3 hours with a standard deviation of about 2.1: on the face of it, almost exactly on the standard. The test will tell us whether the small excess means anything.

### Step 4: Interpret the Graph and Descriptive Statistics

```python
plt.hist(df["Hours_Worked"], bins=6, alpha=0.7, color="green")
plt.title("Weekly Hours Worked")
plt.xlabel("Hours Worked")
plt.ylabel("Frequency")
plt.show()
```

The distribution is tightly clustered around 40 hours and roughly symmetric, with no outliers.

### Step 5: Check for Normality

```python
# Q-Q plot
stats.probplot(df["Hours_Worked"], dist="norm", plot=plt)
plt.title("Q-Q Plot of Hours Worked")
plt.show()

# Shapiro-Wilk test
_, p_shapiro = stats.shapiro(df["Hours_Worked"])
print(f"Shapiro-Wilk p-value: {p_shapiro:.3f}")
```

### Step 6: Transformation or Non-Parametric Test?

The Q-Q plot is close to the line and the Shapiro-Wilk p-value is about 0.46, above 0.05, so the data is compatible with normality. A **one-sample t-test** is appropriate, this time two-sided.

### Step 7: Formulate and Test Hypotheses

```python
# Two-sided one-sample t-test against the 40-hour standard
t_stat, p_value = stats.ttest_1samp(df["Hours_Worked"], 40)
print(f"T-statistic: {t_stat:.3f}, p-value: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: average hours differ from the standard.")
else:
    print("Fail to reject the null hypothesis: no significant deviation from 40 hours.")
```

**Explanation**

The t-statistic is about 0.76 with a two-sided p-value of about 0.45, well above 0.05, so we fail to reject H0: the sample gives no evidence that average weekly hours deviate from the 40-hour standard. The observed mean of 40.3 hours is comfortably within ordinary sampling variation for 25 employees. For HR this is a reassuring null result: workload appears on target, and no intervention is indicated by this data.
